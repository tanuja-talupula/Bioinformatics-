```python
import matplotlib.pyplot as plt

counts = {}
dna = "ATGCGTACGTTAGCTAGCTAACGATCGATCGTACGATCGTAGCTAGCTAGGCTAACG"
for i in dna:
  if i not in counts:
    counts[i] = 1
  else:
    counts[i] += 1
print(counts)

plt.bar(counts.keys(), counts.values())

plt.xlabel("Nucleotide")
plt.ylabel("Count")
plt.title("DNA Nucleotide Frequency")

plt.show()
```
